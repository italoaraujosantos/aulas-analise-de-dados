RESOLVENDO PROBLEMA NA CONTAGEM DE EMPRESAS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_customers = pd.read_csv(
    "customers-10000.csv",
    dtype={"Company": str}
)
contagem_empresas = df_customers.groupby(["Company"]).size()
print(contagem_empresas.to_string())
```

A contagem conta a mesma empresa, com nomes diferentes, como várias empresas diferentes, o que atrapalha nossa análise. Vamos retirar a primeira palavra (nome principal da empresa) e usar esse nome como parâmetro para fazer uma analise mais completa.

```python
df_customers["Empresa_Principal"] = (
    df_customers["Company"]
    .str.strip()
    .str.extract(r'^([A-Za-z]+)', expand=False)
    .str.title()
)

maiores_empresas = df_customers["Empresa_Principal"].value_counts().sort_index()

print(maiores_empresas.to_string())
```
