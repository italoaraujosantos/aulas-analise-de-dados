### Resolvendo a atividade 5 em células

```python
import pandas as pd
import numpy as np

mercados_df = pd.read_csv("vendas_mercados.csv")
print(mercados_df)
```

Resolvendo primeira questão

```python
mercados_df["lucro"] = (
    (mercados_df["Preco_Unitario"] - mercados_df["Custo_Unitario"]) * mercados_df["Quantidade"]
)

lucro_por_mercado = mercados_df.groupby("Mercado")["lucro"].sum()
maior_lucro = lucro_por_mercado.max()

print(lucro_por_mercado.sort_values(ascending=False))
print(f'O maior lucro: {maior_lucro}')
```

Resolvendo questão 2

```python
pagamento_por_regiao = mercados_df.groupby(["Regiao", "Forma_Pagamento"])["ID_Venda"].count()

print(pagamento_por_regiao)
```

RESOLVEDO QUESTÃO 3

```python
vendas_produtos = mercados_df.groupby("Produto")["Quantidade"].sum()

valor_mais_vendido = vendas_produtos.max()
produto_mais_vendido = vendas_produtos.idxmax()
print(f"O produto MAIS vendido: {produto_mais_vendido}"
      f" com {valor_mais_vendido} vendas totais.")

valor_menos_vendido = vendas_produtos.min()
produto_menos_vendido = vendas_produtos.idxmin()
print(f"O produto MENOS vendido: {produto_menos_vendido}"
      f" com {valor_menos_vendido} vendas totais.")
```
